```python
import json
import pathlib
```

```python
outputs_dir = pathlib.Path("outputs")

outputs = []
for file in outputs_dir.iterdir():
    with file.open() as f:
        outputs.append(json.load(f))
```

```python
for output in outputs:
    result = output[0]["results"][0]
    if result["times"] == []:
        continue

    print(f"{result["problem"]["meta"]["name"]} | {result["total_time"]}")
```

```python
import pandas as pd
```

```python
time_data = []
for output in outputs:
    for result in output[0]["results"]:
        if result["times"]:
            d = {"name": result["problem"]["meta"]["name"], "total_time": result["total_time"]}
            for time in result["times"]:
                d[time["name"]] = time["time"]

            d["heuristic"] = result["args"][-1]

            time_data.append(d)
        else:
            param_runs = result["param_runs"]
            for run in param_runs:
                d = {"name": f"{result["problem"]["meta"]["name"]} {run["name"]}", "total_time": run["total_time"]}
                for time in run["times"]:
                    if time["name"].startswith("Replaced"):
                        continue

                    d[time["name"]] = time["time"]

                d["heuristic"] = result["args"][-1]

                time_data.append(d)
```

```python
df = pd.DataFrame(time_data)
```

```python
df.sort_values("total_time", ascending=False)
```
